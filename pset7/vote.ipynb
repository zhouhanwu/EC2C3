```python
import pandas as pd
import numpy as np
import pyfixest as pf

from pathlib import Path
```

```python
ROOT = Path.cwd()
vote_df = pd.read_csv(ROOT / 'vote.csv')
vote_df.dtypes
```

```python
vote_df.describe()
```

```python
fit = pf.feols('demo_share ~ distance', data=vote_df, vcov='hetero')
fit.summary()
```

```python
# fillna
vote_df['demo_share'] = vote_df['demo_share'].fillna(100 - vote_df['rep_share'] - vote_df['other_share'])
```

```python
fit = pf.feols('demo_share ~ distance', data=vote_df, vcov='hetero')
fit.summary()
```

```python
restricted_df = vote_df.loc[vote_df['distance'] >= 5]
```

```python
fit = pf.feols('demo_share ~ distance', data=restricted_df, vcov='hetero')
fit.summary()
```

```python
fit = pf.feols('demo_share ~ distance + black', data=vote_df, vcov='hetero')
fit.summary()
```
