```python
import pandas as pd
import numpy as np
import pyfixest as pf

from pathlib import Path
```

```python
ROOT = Path.cwd()
wellbeing_df = pd.read_csv(ROOT / 'us_wellbeing_survey.csv')
wellbeing_df.dtypes
```

```python
wellbeing_df.shape
```

```python
wellbeing_df.describe()
```

```python
fit = pf.feols("mh_conditions ~ np.log(sleephrs)", data=wellbeing_df)
fit.summary()
```

```python
fit = pf.feols("mh_conditions ~ np.log(sleephrs) + pain_back_joint", data=wellbeing_df)
fit.summary()
```

```python
fit = pf.feols("mh_conditions ~ pain_back_joint", data=wellbeing_df)
wellbeing_df['mh_resid'] = fit.resid()
```

```python
fit = pf.feols("np.log(sleephrs) ~ pain_back_joint", data=wellbeing_df)
wellbeing_df['sleep_resid'] = fit.resid()
```

```python
fit = pf.feols("mh_resid ~ sleep_resid", data=wellbeing_df)
fit.summary()
```

```python
fit = pf.feols("mh_conditions ~ sleep_resid", data=wellbeing_df)
fit.summary()
```

```python
fit = pf.feols("mh_conditions ~ np.log(sleephrs) + pain_back_joint + physical_activity", data=wellbeing_df)
fit.summary()
```

```python
R = np.array([[0, 0.1, 0, 1]])
fit.wald_test(R=R, q=0)
```

```python
R = np.array([[0, 0.1, 0, 1],     # b3 + 0.1·b1 = 0
              [0, 1.2, 1, 0]])    # b2 + 1.2·b1 = 0
q = np.array([0.0, 0.0])

res = fit.wald_test(R=R, q=q)
```

```python
from scipy import stats
F = res["statistic"] / 2
dfd = fit._dfd              # residual degrees of freedom (n − k)
p = stats.f.sf(F, 2, dfd)
```
